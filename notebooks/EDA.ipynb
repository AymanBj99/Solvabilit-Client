```python
import seaborn as sns
import matplotlib.pyplot as plt

# Distribution de l'âge
sns.histplot(dt['AGE'], bins=30, kde=True)
plt.title("Distribution de l'âge")
plt.show()

#distribution de sexe
sns.countplot(x='SEX', hue='default.payment.next.month', data=dt, palette={0:'blue', 1:'orange'})
plt.title("Sexe et défaut de paiement")
plt.xlabel("Sexe (1=Homme, 2=Femme)")
plt.ylabel("Nombre de clients")
plt.legend(title="Solvabilité", labels=["0 = solvable", "1 = non solvable"])
plt.show()


#Education vs défaut de paiement
sns.countplot(x='EDUCATION', hue="default.payment.next.month", data=dt)
plt.title("Education et défaut de paiement")
plt.show()

# Statut marital vs défaut de paiement
sns.countplot(x='MARRIAGE', hue='default.payment.next.month', data=dt)
plt.title("Statut marital et défaut de paiement")
plt.xlabel("Statut marital (1=Married, 2=Single, 3=Others)")
plt.ylabel("Nombre de clients")
plt.show()

# Calcul du taux de défaut par statut marital
marriage_default = dt.groupby("MARRIAGE")["default.payment.next.month"].mean()
print("Taux de défaut (%) par statut marital :")
print((marriage_default * 100).round(2))

# Distribution de PAY_0
sns.countplot(x='PAY_0', hue='default.payment.next.month', data=dt)
plt.title("PAY_0 (statut remboursement septembre) vs défaut de paiement")
plt.xlabel("Statut de remboursement (-1 = à temps, 1+ = retard en mois)")
plt.ylabel("Nombre de clients")
plt.show()


pay_features = ['PAY_0', 'PAY_2', 'PAY_3', 'PAY_4', 'PAY_5', 'PAY_6']

for col in pay_features:
    plt.figure(figsize=(7,5))
    sns.countplot(x=col, hue='default.payment.next.month', data=dt)
    plt.title(f"{col} vs défaut de paiement")
    plt.xlabel("Statut de remboursement (-1 = à temps, 0 = payé, 1+ = retard en mois)")
    plt.ylabel("Nombre de clients")
    plt.show()

#Matrice de corrélation
plt.figure(figsize=(12,10))
sns.heatmap(dt.corr(), annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Matrice de corrélation")
plt.show()




```
